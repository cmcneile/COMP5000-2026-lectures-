## Towards the creating the program

```python
name = "fit.txt"
f=open(name, "r")  
tmp = f.readlines()
no_pressup = 0
for x in tmp:
  xx = x.rstrip()
#  print(xx)
  if "Pressup" in xx :
#     print(xx)
     xx_split = xx.split()
#     print(xx_split)
     no_pressup = no_pressup + int(xx_split[1])
f.close()
print("Number of pressups = ", no_pressup)
```

##  Final solution

```python
name = "fit.txt"
f=open(name, "r")  
tmp = f.readlines()
no_pressup = 0
for x in tmp:
  xx = x.rstrip()
  if "Pressup" in xx :
    print(xx)
    xx_split = xx.split()
    no_pressup += int(xx_split[1])
f.close()
print(xx_split[1])
print("Number of pressups = ", no_pressup)
```

# Python code to read csv file

We could use the string handling tools in Python
to read this type of file. But there is already a module
called csv https://docs.python.org/3/library/csv.html,
which will do most of the work.

## Example of reading a csv file
* You need to download the csv file https://raw.githubusercontent.com/cmcneile/COMP5000-2022-lectures/main/fit.csv
* To load a csv file the csv module can be used (https://www.programiz.com/python-programming/csv)

```python
import csv
name = "fit.csv"
with open(name) as csvfile:
    readCSV = csv.reader(csvfile, delimiter=',')
    next(readCSV, None)  # skip the headers
    for row in readCSV:
      print(row)
```

##  Solving the problem if the data is stored in a csv file

```python
import csv
name = "fit.csv"
total = 0
with open(name) as csvfile:
    readCSV = csv.reader(csvfile, delimiter=',')
    next(readCSV, None)  # skip the headers
    for row in readCSV:
      total += int( row[1] )
print("Number of pressups = ", total)
```
